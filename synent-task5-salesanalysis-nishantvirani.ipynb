 # Task 5: Sales Data Analysis

##Problem Statement

- The objective of this project is to analyze business performance using the Superstore Sales dataset.

##Dataset Details :

- The dataset includes sales, profits, products, customers, and regional information.

##Approach :

- I analyzed monthly revenue trends, top-selling products, and profit patterns using data visualization techniques.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("train.csv")

df.info()

df.describe()

print(df.shape)

df.isnull().sum()

print(df.head())

print(df.columns)

print(df.isnull().sum())
# convert ORDER DATE object into OEDER DATE
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')

```

##Monthly revenue trends :

- This visualization shows how sales revenue changes over time and helps identify seasonal patterns and business growth.

```python
import matplotlib.pyplot as plt

monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()
print(monthly_sales)

monthly_sales.plot(kind='bar')

plt.title("Monthly Revenue Trends")
plt.ylabel('Revenue')
plt.xlabel('Month')
plt.show()
```

##Top Selling Products

- This analysis identifies the products that contribute the highest sales revenue.

```python
import matplotlib.pyplot as plt

#Top-selling products contribute major revenue

top_selling_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_selling_products)

top_selling_products.plot(kind='bar')

plt.title("Top 10 Selling products")
plt.ylabel("Revenue")
plt.xlabel("Products Name")
plt.show()

```

##Profit analysis

- Technology/Furniture category generates       significant sales.


- Certain regions outperform others in revenue generation.

- Consumer segment contributes the largest share of sales.

```python
import matplotlib.pyplot as plt

#Technology/Furniture category generates significant sales

category_sales = df.groupby('Category')['Sales'].sum()

category_sales.plot(kind='bar')

plt.title("Sales by Category")
plt.ylabel("Revenue")
plt.xlabel("Category")

plt.show()

#Certain regions outperform others in revenue generation
regions_sales = df.groupby('Region')['Sales'].sum()

regions_sales.plot(kind='bar')
plt.title("Sales by Region")
plt.ylabel("Revenue")
plt.xlabel("Region")
plt.show()

#Consumer segment contributes the largest share of sales.
segment_sales = df.groupby('Segment')['Sales'].sum()

segment_sales.plot(kind='bar')
plt.title("Sales by Segment")
plt.ylabel("Revenue")
plt.xlabel("Segment")
plt.show()
```

## Business Insights

- Sales fluctuate over different months.
- Top-selling products contribute significantly to revenue.
- Some categories perform better than others.
- Regional performance varies across the business.
